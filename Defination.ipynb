### Types of Recommendatuons Systems
1. Content-Based Filtering
2. Collaborative Filtering
3. Market Basket Analysis, etc.

1. Content-Based Filtering
- this method recommends items that are similar to those the user has liked in the past.
- Example:
- A user frequently watches science-fiction movies.
-the system recommends other science-fiction movies with similar features.


2. Collaborative Filtering
- this method recommends items based on the preferences of user with similar interests.
- Example
- User A and user B bith like product X and products Y .
- if User A also like product Z, the system may recommend product Z to User B.

3. Market Basket Analysis
- this approach identifies products that are frequently purchased together.
-Example:
- Customer who Buy a laptop often buy a mouse and Keyboard.
- the System recommends these related products  during checkout.

**Cosine Similarity**:   
cosine similarity is a popular measure used to determine how similar two users or items are:
 >-  $\text{cosine Similarity}=  \frac{A \cdot B}{||A||.||B||}$
  >- A.B = Dot product of vectors A and B
 > -  ||A|| = Magnitude of Vector A
 >-  ||B|| = Magnitude of vector B.

** Example**:
Suppose two user rate three movies as:
- User A = [5,4,0]
- user B = [4,5,1]

1. Dot Product:
- $(5 \times 4) + (4 \times 5) + (0 \times 1 ) = 40$

2. magnitude of User A:
 - $\sqrt{5^2 + 4^2 + 0^2} = 6.40$

3. Magnitude of B:
 - $\sqrt{4^2 + 5^2 + 1^2} = 6.48$

4. Cosine Similarity:
 - $\frac{40}{(6.40 \times 6.48)} = 0.96$
 - Since the similarity score is closer to 1, the two have very similar preferences.
